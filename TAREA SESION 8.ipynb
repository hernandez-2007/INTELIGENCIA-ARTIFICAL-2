 # Modelado Predictivo del Costo de Casas (Regresión Lineal vs. Árboles de Decisión)

Este notebook tiene como objetivo construir un modelo predictivo para estimar el costo de las casas o propiedades. Compararemos el rendimiento de dos algoritmos de regresión populares: Regresión Lineal y Árboles de Decisión.  ## 1. Configuración Inicial y Carga del Dataset

Primero, importaremos las librerías necesarias y cargaremos el dataset de precios de casas de California, que es un dataset estándar para problemas de regresión.

```python
# Importamos la librería pandas para la manipulación y análisis de datos en formato de DataFrame.
import pandas as pd

# Importamos el dataset de precios de casas de California de scikit-learn.
# Este dataset es ideal para problemas de regresión y ya viene preprocesado.
from sklearn.datasets import fetch_california_housing

# Cargamos el dataset de California Housing. 'as_frame=True' lo convierte directamente en un DataFrame de pandas.
housing = fetch_california_housing(as_frame=True)

# Creamos un DataFrame principal combinando las características (data) y la variable objetivo (target).
# La variable objetivo (median house value) es lo que queremos predecir.
df = pd.DataFrame(data=housing.data, columns=housing.feature_names)
# Añadimos la columna del valor objetivo al DataFrame para tener todo junto y facilitar el análisis.
df['MedHouseVal'] = housing.target

# Mostramos las primeras 5 filas del DataFrame para tener una idea de la estructura de los datos.
# Esto nos permite ver las características (variables independientes) y la variable objetivo (MedHouseVal).
display(df.head())
```
