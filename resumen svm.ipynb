Las Máquinas de Vectores de Soporte (SVM por sus siglas en inglés, Support Vector Machines) son algoritmos de aprendizaje supervisado utilizados tanto para problemas de clasificación como de regresión. Funcionan encontrando el hiperplano óptimo que mejor separa las diferentes clases de datos en un espacio de características. En el caso más simple, para datos linealmente separables, el hiperplano es una línea (en 2D) o un plano (en 3D) que maximiza el margen entre las clases, es decir, la distancia a los puntos de datos más cercanos de cada clase (conocidos como vectores de soporte).

Para datos que no son linealmente separables, las SVM utilizan una técnica llamada "truco del kernel". Esto implica mapear los datos a un espacio de dimensiones superiores donde pueden ser linealmente separables, sin necesidad de calcular explícitamente las coordenadas en ese espacio de alta dimensión. Los kernels más comunes incluyen el lineal, el polinomial y el de función de base radial (RBF).

Características clave de las SVM:

Margen Máximo: Buscan el hiperplano que maximiza la distancia a los puntos más cercanos de cada clase, lo que ayuda a una mejor generalización.
Vectores de Soporte: Solo un subconjunto de los puntos de entrenamiento (los más cercanos al hiperplano o que cruzan el margen) son cruciales para definir el modelo. Estos son los vectores de soporte.
Efectividad en Espacios de Alta Dimensión: Son muy efectivas en espacios de dimensiones elevadas.
Versatilidad con Funciones Kernel: La elección de una función kernel permite adaptar la SVM a diferentes tipos de distribuciones de datos.
Menos propensas al sobreajuste: Debido a su enfoque en el margen máximo, tienden a ser menos propensas al sobreajuste que otros modelos en ciertos escenarios.
En resumen, las SVM son herramientas potentes y versátiles para la clasificación y regresión, conocidas por su solidez y capacidad para manejar datos complejos mediante el uso de kernels.
